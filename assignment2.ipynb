### Q1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10 numbers in list now).

##### i. Print L

```python
roll_no = 1024170105

l = [] # creating a list

while roll_no > 0 :
  dig = roll_no % 10
  l.append(dig*10)

  roll_no = roll_no//10 # rememeber here division is using //

l.reverse()
print(l)
```

### Q3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

##### i.Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
import random
from collections import Counter

roll_no = 1024170105
random.seed(roll_no)

random_numbers = [random.randint(100, 900) for _ in range(100)]  # a list for storing the random numbers

print("Random numbers:")
print(random_numbers)
```

##### ii. Count and print all odd numbers in the list.

```python
odd_numbers_list = [i for i in random_numbers if i % 2 != 0]

print("Odd numbers:")
print(odd_numbers_list)
```

##### iii. Count and print all even numbers in the list.

```python
even_numbers_list = [i for i in random_numbers if i % 2 == 0]

print("Even numbers:")
print(even_numbers_list)
```

##### iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers
found using list comprehension.

```python
def is_prime(n):
  if n < 2 :
    return False
  for i in range(2 , n):
    if n % i == 0 :
      return False

  return True

prime_numbers_list = [i for i in random_numbers if is_prime(i)]

print("Prime numbers:")
print(prime_numbers_list)
```

##### v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
frequency = Counter(random_numbers)

# this frequency works like a dictionary where the key is the number in the list
# and the value of the key is the how much times the number is appearing in the list

most_frequent_number , frequency_count = frequency.most_common(1)[0]  # (1) means give only 1 item

print("Most frequent number:", most_frequent_number)
print("Frequency count:", frequency_count)
```

### Q4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

##### vi. Find the union of A and B — all unique values across both sets.


```python
roll_no = 1024170105

digits = []

temp = roll_no

while temp > 0 :
  dig = temp % 10
  digits.append(dig)
  temp = temp // 10

digits.reverse()

A = {i*7 for i in digits}
B = {i*9 for i in digits}

print("Set A:", A)
print("Set B:", B)

find_union = A.union(B) ;

print(find_union)
```

##### vii. Find the intersection of A and B — values common to both

```python
find_intersection = A.intersection(B)
print(find_intersection)
```

##### viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference().

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements present in one set but not the other,
# while symmetric_difference() gives elements that are in either set but not both.
```

##### ix. Find the symmetric difference of A and B.

```python
symmetric = A.symmetric_difference(B)

print("Symmetric difference:", symmetric)
```
