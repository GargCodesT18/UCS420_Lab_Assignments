Q1.  Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.).
1. Convert text to lowercase and remove punctua on using re.
2. Tokenize the text into words and sentences.
3. Split using split() and word_tokenize() and compare how Python split and NLTK’s word_tokenize() differ.
4. Remove stopwords (using NLTK's stopwords list).
5. Display word frequency distribu on (excluding stopwords).

```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from collections import Counter
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """
Artificial intelligence is one of my favorite topics because it combines
computer science, mathematics, and creativity. I enjoy learning how machines
can understand data, recognize patterns, and make useful predictions.
Artificial intelligence is becoming important in healthcare, education,
robotics, finance, and many other fields.
"""
```

```python
text_lower = text.lower()

text_clean = text_lower.translate(
    str.maketrans('', '', string.punctuation)
)

print("Cleaned Text:")
print(text_clean)
```

```python
tokens = word_tokenize(text_clean)

print("Tokens using word_tokenize():")
print(tokens)
```

```python
split_tokens = text_clean.split()

print("Using Python split():")
print(split_tokens)

print("Using NLTK word_tokenize():")
print(tokens)

print("Number of tokens using split():", len(split_tokens))
print("Number of tokens using word_tokenize():", len(tokens))
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print("Tokens after removing stopwords:")
print(filtered_tokens)
```

```python
word_freq = Counter(filtered_tokens)

print("Word Frequency:")
for word, frequency in word_freq.most_common():
    print(word, ":", frequency)

words = list(word_freq.keys())
frequencies = list(word_freq.values())

plt.figure(figsize=(10, 5))
plt.bar(words, frequencies)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Word Frequency Distribution")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Q2. Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall()
2. Remove stop words using NLTK’s stopword list
3. Perform stemming with PorterStemmer
4. Perform lemma za on with WordNetLemma zer
5. Compare the stemmed and lemmatized outputs and explain when you’d prefer one over the other.

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
tokens = word_tokenize(text.lower())

# Keep alphabetic words only
alpha_tokens = [
    word for word in tokens
    if word.isalpha()
]
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in alpha_tokens
    if word not in stop_words
]

print("Filtered Tokens:")
print(filtered_tokens)
```

```python
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_tokens
]

print("Stemmed Words:")
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]

print("Lemmatized Words:")
print(lemmatized_words)
```

```python
for original, stemmed, lemmatized in zip(
    filtered_tokens,
    stemmed_words,
    lemmatized_words
):
    print(
        f"{original:15} -> "
        f"Stem: {stemmed:15} "
        f"Lemma: {lemmatized}"
    )
```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representa on.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
texts = [
    "Artificial intelligence is transforming modern technology and healthcare.",
    "Machine learning helps computers learn patterns from data.",
    "Deep learning uses neural networks for image and language applications."
]
```

```python
count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Matrix:")
print(bow_matrix.toarray())

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("TF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
for i, row in enumerate(tfidf_matrix.toarray()):

    top_indices = row.argsort()[-3:][::-1]

    top_words = [
        (feature_names[index], row[index])
        for index in top_indices
    ]

    print(f"Text {i + 1}:")

    for word, score in top_words:
        print(f"{word} -> {score:.4f}")
```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:

a. Jaccard Similarity using sets

b. Cosine Similarity using TfidfVectorizer + cosine_similarity()

c. Analyze which similarity metric gives be er insights in your case.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
text1 = """
Artificial intelligence enables computers to learn from data,
recognize patterns, and make intelligent decisions.
"""

text2 = """
Blockchain is a distributed technology that securely stores
transactions using decentralized networks.
"""
```

```python
words1 = set(text1.lower().split())
words2 = set(text2.lower().split())

print("Text 1 words:")
print(words1)

print("Text 2 words:")
print(words2)
```

```python
intersection = words1.intersection(words2)
union = words1.union(words2)

jaccard_similarity = len(intersection) / len(union)

print("Jaccard Similarity:")
print(jaccard_similarity)
```

```python
texts = [text1, text2]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(texts)

cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("Cosine Similarity:")
print(cosine_sim[0][0])
```

```python
print("Analysis:")
print("""
Jaccard similarity compares the overlap between the unique words
of the two texts. It is simple and useful when only the presence
or absence of words is important.

Cosine similarity using TF-IDF considers the importance of words
and compares the direction of the document vectors. Therefore,
cosine similarity generally provides more useful information for
text similarity when word importance is relevant.
""")
```

Q5. Write a short review for a product or service.
1. Use TextBlob or VADER to find polarity & subjec vity for each review.
2. Classify reviews into Posi ve / Nega ve / Neutral.
3. Create a word cloud using the wordcloud library for all posi ve reviews.

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from wordcloud import WordCloud
import matplotlib.pyplot as plt

nltk.download('vader_lexicon')
```

```python
reviews = [
    "This product is absolutely amazing and works perfectly.",
    "I really love this product. The quality is excellent.",
    "The service was okay, nothing special.",
    "This product is terrible and completely useless.",
    "I am very disappointed with the poor quality.",
    "The product is good and I am happy with my purchase."
]
```

```python
sia = SentimentIntensityAnalyzer()
```

```python
positive_reviews = []

for review in reviews:

    scores = sia.polarity_scores(review)

    compound = scores['compound']

    if compound >= 0.05:
        sentiment = "Positive"
        positive_reviews.append(review)

    elif compound <= -0.05:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    print("Review:")
    print(review)

    print("Positive Score:", scores['pos'])
    print("Negative Score:", scores['neg'])
    print("Neutral Score:", scores['neu'])
    print("Compound Score:", scores['compound'])
    print("Sentiment:", sentiment)
```

```python
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

Q6. Choose your own paragraph (~100 words) as training data.
1. Tokenize text using Tokenizer() from keras.preprocessing.text
2. Create input sequences and build a simple LSTM or Dense model
3. Train the model and generate 2–3 new lines of text star ng from any seed word you provide.

```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
text = """
Artificial intelligence is changing the way people use technology.
Machine learning allows computers to learn patterns from data and
make useful predictions. Deep learning uses neural networks to solve
complex problems in vision, speech, and language. Artificial intelligence
is used in healthcare, education, finance, robotics, and transportation.
Students can learn artificial intelligence by studying mathematics,
programming, algorithms, and data science. Building practical projects
helps students understand machine learning concepts. Artificial intelligence
will continue to influence many industries and create new opportunities
for researchers, engineers, and developers.
"""
```

```python
text = text.lower()
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total words in vocabulary:", total_words)
```

```python
token_list = tokenizer.texts_to_sequences([text])[0]

print(token_list)
```

```python
input_sequences = []

for i in range(1, len(token_list)):

    sequence = token_list[:i + 1]

    input_sequences.append(sequence)

max_sequence_len = max(
    [len(sequence) for sequence in input_sequences]
)
```

```python
input_sequences = np.array(
    pad_sequences(
        input_sequences,
        maxlen=max_sequence_len,
        padding='pre'
    )
)

X = input_sequences[:, :-1]

y = input_sequences[:, -1]

y = tf.keras.utils.to_categorical(
    y,
    num_classes=total_words
)


print("\nInput shape:", X.shape)
print("Output shape:", y.shape)
```
