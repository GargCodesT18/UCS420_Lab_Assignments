Q.2 Questions on Basic NumPy Array:

a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

b) Find the most frequent value and their indice(s) in the following arrays:

i. x = np.array([1,2,3,4,5,1,2,1,1,1])

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])


```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

print("Array:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", counts[np.argmax(counts)])
```

```python
y = np.array([1, 2, 3, 4, 2, 4, 3, 3, 1])

values, counts = np.unique(y, return_counts=True)

most_frequent = values[np.argmax(counts)]

print("Array:", y)
print("Most frequent value:", most_frequent)
print("Frequency:", counts[np.argmax(counts)])
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?


```python
my_array = np.linspace(10, 100, 25)

print("Array:")
print(my_array)

print("Dimensions:", my_array.ndim)
print("Shape:", my_array.shape)
print("Total elements:", my_array.size)
print("Data type:", my_array.dtype)
print("Total bytes:", my_array.nbytes)
```

```python
print("Transpose:")
print(my_array.T)
```

```python
reshaped_array = my_array.reshape(5, 5)

print("Reshaped array:")
print(reshaped_array)

print("Dimensions after reshape:", reshaped_array.ndim)
print("Shape after reshape:", reshaped_array.shape)
```

```python
print("Transpose of reshaped array:")
print(reshaped_array.T)
```
