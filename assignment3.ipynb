### Q.6 Create a sample dataset (employees.csv) containing information about employees in a company.
### WAP to determine the following using dataset in Q6:

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

print("Employee DataFrame:")
print(df)
```

##### a) Shape (number of rows and columns) of the DataFrame.

```python
print(df.shape)
```

##### b) Summary of the DataFrame that includes the data types and non-null counts for each column.

```python
df.info()
```

##### c) Generate descriptive statistics.

```python
print(df.describe())
```

##### d) Display the first 5 rows and last 3 rows of the dataset.

```python
print(df.head(5))
print(df.tail(3))
```

##### e) Calculate the following statistics from the dataset:

###### i. Average salary of employees

```python
average_salary = df["Salary"].mean()
print("Average Salary:", average_salary)
```

###### ii. Total bonus paid to all employees

```python
total_bonus = df["Bonus"].sum()
print("Total Bonus Paid:", total_bonus)
```

###### iii. The youngest employee's age.

```python
youngest_age = df["Age"].min()
print("Youngest Employee's Age:", youngest_age)
```

###### iv. The highest performance rating.

```python
highest_rating = df["Rating"].max()
print("Highest Performance Rating:", highest_rating)
```

##### f) Sort the DataFrame by the Salary column in descending order.

```python
df = df.sort_values(by="Salary", ascending=False)
print(df)
```

##### g) Add a new column that categorizes employees based on their performance rating:

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"


df["Performance_Category"] = df["Rating"].apply(performance_category)

print("DataFrame with Performance Category:")
print(df)
```

###### h) Identify missing values

```python
print(df.isnull().sum())
```

##### i) Rename Employee_ID column to ID

```python
df = df.rename(columns={"Employee_ID": "ID"})

print(df)
```

##### j) Find employees based on conditions

###### i. Employees having more than 5 years of experience

```python
experienced_employees = df[df["Years_of_Experience"] > 5]
print(experienced_employees)
```

###### ii. Employees belonging to IT department

```python
it_employees = df[df["Department"] == "IT"]
print(it_employees)
```

##### k) Add Tax column which deducts 10% of Salary

```python
df["Tax"] = df["Salary"] * 0.10
print(df)
```

##### l) Save modified DataFrame to a new CSV file

```python
df.to_csv("employees_modified.csv", index=False)
```
