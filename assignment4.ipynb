Q1: Build Your Personalized Knowledge Base: Take your college roll number. Extract its digits. Build a pandas
DataFrame with exactly 6 FAQ entries: 4 fixed entries (given in table below) and other 2 entries constructed from your
own roll number digits as follows:

1. Take the LAST TWO DIGITS of your roll number. For each digit d, compute category = ["billing", "account",
"general"][d % 3]. Invent one realistic question+answer+3 keywords per entry that fits the assigned category (e.g.
if d%3 gives "account", write a question like “how do I update my registered mobile number”).

2. Example roll number ...23 -> digits 2, 3
3.  digit 2 -> category[2 % 3] = general
4.  digit 3 -> category[3 % 3] = billing

```python
roll_number = input("Enter your roll number: ")

# Take the last two digits
digit1 = int(roll_number[-2])
digit2 = int(roll_number[-1])

print("Last two digits:", digit1, digit2)
```

```python
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": ["fee cost", "price charge", "annual"],
        "category": "billing"
    },

    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": ["password reset", "login", "forgot"],
        "category": "account"
    },

    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": ["hours", "timing", "open time"],
        "category": "general"
    },

    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": ["payment", "pay", "fee"],
        "category": "billing"
    }
]
```

```python
categories = ["billing", "account", "general"]

category1 = categories[digit1 % 3]
category2 = categories[digit2 % 3]

entry5 = {
    "question": "how do I update my account details",
    "answer": "You can update your account details from the Settings section.",
    "keywords": ["account", "details", "update"],
    "category": category1
}

entry6 = {
    "question": "how can I get help with my service",
    "answer": "You can contact customer support for help with the service.",
    "keywords": ["help", "support", "service"],
    "category": category2
}

fixed_entries.append(entry5)
fixed_entries.append(entry6)
```

```python
faq = pd.DataFrame(fixed_entries)

print(faq)
```

Q2: Generate and Score a Hypothesis - Implement a scoring function that takes a query string and returns all matching
entries ranked by confidence

```python
def search_faq(query):
    query = query.lower()
    results = []

    for i in range(len(faq)):
        text = faq.loc[i, "question"].lower()

        for keyword in faq.loc[i, "keywords"]:
            text = text + " " + keyword.lower()

        words = query.split()
        matches = 0

        for word in words:
            if word in text:
                matches += 1

        if matches > 0:
            confidence = matches / len(words)

            results.append({
                "question": faq.loc[i, "question"],
                "answer": faq.loc[i, "answer"],
                "category": faq.loc[i, "category"],
                "confidence": confidence
            })

    results = sorted(results, key=lambda x: x["confidence"], reverse=True)

    return results
```

```python
query = input("Enter your question: ")

results = search_faq(query)

for result in results:
    print("\nQuestion:", result["question"])
    print("Answer:", result["answer"])
    print("Category:", result["category"])
    print("Confidence:", result["confidence"])
```

Q3: Write a function same_category(category_name, df) that returns all questions belonging to a given category. Call it
using the category of one of the personalized entries from Q1, and print the result.

```python
def same_category(category_name, df):
    result = df[df["category"] == category_name]
    return result
```

Q6: Modify your Q2 scoring function so that if two or more entries tie for the highest score, it does not silently pick one
— it prints all matching entries instead, so the user can see every equally good match. Demonstrate with one query that
produces a tie (e.g. a query matching both "fee" entries) and one that doesn't.

```python
def search_faq_with_ties(query):
    query = query.lower()
    results = []

    for i in range(len(faq)):
        text = faq.loc[i, "question"].lower()

        for keyword in faq.loc[i, "keywords"]:
            text = text + " " + keyword.lower()

        words = query.split()
        matches = 0

        for word in words:
            if word in text:
                matches += 1

        if matches > 0:
            confidence = matches / len(words)

            results.append({
                "question": faq.loc[i, "question"],
                "answer": faq.loc[i, "answer"],
                "category": faq.loc[i, "category"],
                "confidence": confidence
            })

    if len(results) == 0:
        print("No matching FAQ found.")
        return

    results.sort(key=lambda x: x["confidence"], reverse=True)

    highest_score = results[0]["confidence"]

    print("Highest confidence:", highest_score)
    print("\nMatching entries:")

    for result in results:
        if result["confidence"] == highest_score:
            print("\nQuestion:", result["question"])
            print("Answer:", result["answer"])
            print("Category:", result["category"])
            print("Confidence:", result["confidence"])
```

```python
search_faq_with_ties("fee")
```
