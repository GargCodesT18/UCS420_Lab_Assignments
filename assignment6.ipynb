Q1. Consider the following sensor readings:
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:

a. Add 2°C to every reading using vectorization.

b. Convert all temperatures from Celsius to Fahrenheit using:
F = 9
5 C + 32

c. Identify all readings greater than 32°C using Boolean indexing.

d. Count how many readings exceed 32°C.

e. Explain how vectorization and Boolean indexing can make data processing more efficient
than explicitly iterating through every value

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

print("Original temperatures:", temperature)
```

```python
corrected_temperature = temperature + 2

print("Temperature after adding 2°C:")
print(corrected_temperature)
```

```python
fahrenheit = (9 / 5) * corrected_temperature + 32

print("Temperature in Fahrenheit:")
print(fahrenheit)
```

```python
greater_than_32 = corrected_temperature > 32

print("Readings greater than 32°C:")
print(corrected_temperature[greater_than_32])
```

```python
count = np.sum(corrected_temperature > 32)

print("Number of readings greater than 32°C:", count)
```

```python
print("Vectorization performs operations on the whole array at once,")
print("so we do not need a loop for every element.")

print("Boolean indexing uses True/False conditions to select")
print("only the elements that satisfy a condition.")

print("Both methods make NumPy code shorter, faster and easier to read.")
```

Q2. The following matrix represents the number of daily steps recorded by four users
over three days:

steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])

Perform the following:

a. Calculate the total steps recorded.

b. Calculate the mean number of steps.

c. Find the maximum and minimum values.

d. Calculate the total steps for each day using axis=0.

e. Calculate the total steps for each user using axis=1.

f. Find the user/day position containing the maximum number of steps using argmax().

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(steps)
```

```python
total_steps = np.sum(steps)

print("Total steps:", total_steps)
```

```python
mean_steps = np.mean(steps)

print("Mean number of steps:", mean_steps)
```

```python
maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum steps:", maximum)
print("Minimum steps:", minimum)
```

```python
user_total = np.sum(steps, axis=1)

print("Total steps for each user:")
print(user_total)
```

```python
day_total = np.sum(steps, axis=0)

print("Total steps for each day:")
print(day_total)
```

```python
position = np.unravel_index(np.argmax(steps), steps.shape)

print("Position of maximum steps:", position)
print("Maximum steps:", steps[position])
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based
on three factors:
• Sleep hours
• Activity level
• Stress level
Consider:
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])
y = np.array([40, 65, 30, 85])
Here, each row represents one user and each column represents a feature.
Perform the following:

a. Display the shape and dimensions of X.

b. Calculate: X.T and explain what the transpose represents.

c. Calculate the matrix product: X.T @ X

d. Calculate the matrix inverse using: np.linalg.inv() where appropriate.

e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋!𝑋)"#𝑋!𝑦 using NumPy.

f. Explain what the resulting coefficients represent in the context of the three user
features.

g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to
obtain a predicted assistance score.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("X:")
print(X)

print("y:")
print(y)
```

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
XT = X.T

print("Transpose of X:")
print(XT)
```

```python
XT_X = X.T @ X

print("X.T @ X:")
print(XT_X)
```

```python
inverse = np.linalg.inv(XT_X)

print("Inverse of X.T @ X:")
print(inverse)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("OLS coefficients:")
print(beta)
```

```python
print("The first coefficient represents the effect of sleep hours.")
print("The second coefficient represents the effect of activity level.")
print("The third coefficient represents the effect of stress level.")
```

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
